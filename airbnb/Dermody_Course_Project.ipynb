```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, MinMaxScaler
import matplotlib.pyplot as plt
```

```python
#Import the data file
df = pd.read_csv("listings1.csv")
df.info()
df.head()
```

```python
#Check for missing values
for c in df.columns:
    miss = df[c].isnull().sum()
    if miss!=0:
        print("{} has {} missing value(s)".format(c,miss))
```

```python
#Drop missing values
df = df.dropna()
df.info()
```

```python
#Reformat columns
df['host_response_rate'] = df['host_response_rate'].str.rstrip('%').astype(float).div(100)
df['host_acceptance_rate'] = df['host_acceptance_rate'].str.rstrip('%').astype(float).div(100)
df['host_is_superhost'] = df['host_is_superhost'].map({'t': 1, 'f': 0})
df['host_has_profile_pic'] = df['host_has_profile_pic'].map({'t': 1, 'f': 0})
df['host_identity_verified'] = df['host_identity_verified'].map({'t': 1, 'f': 0})
df['price'] = (df['price'].str.strip().str.replace(r'[\$,]', '', regex=True).astype(float))
df['has_availability'] = df['has_availability'].map({'t': 1, 'f': 0})
df['instant_bookable'] = df['instant_bookable'].map({'t': 1, 'f': 0})
```

```python
#Remove price outliers
df = df[df['price'] <= 2000]
```

```python
#Export file
df.to_csv('listings_preprocessed.csv')
```

```python
# Summary Statistics for relevant variables
var = [
    'estimated_revenue_l365d',
    'estimated_occupancy_l365d',
    'price',
    'minimum_nights',
    'accommodates',
    'bedrooms',
    'beds',
    'host_is_superhost',
    'host_response_rate',
    'host_acceptance_rate',
    'number_of_reviews',
    'review_scores_rating',
    'instant_bookable',
    'host_listings_count'
]

summary_stats = df[var].describe().T[['mean', 'std', 'min', 'max']]
print(summary_stats)
```

```python
# --- LOG TRANSFORMATIONS ---

# Log-transform revenue to reduce skew
df['log_revenue'] = np.log1p(df['estimated_revenue_l365d'])

# Log-transform price
df['log_price'] = np.log1p(df['price'])

# Log-transform number of reviews
df['log_reviews'] = np.log1p(df['number_of_reviews'])

# --- NEIGHBORHOOD DUMMIES ---

# Identify top 5 neighborhoods by count
top_neigh = df['neighbourhood_cleansed'].value_counts().nlargest(5).index

# Keep top 5 as separate dummies, group the rest as "Other"
df['neighbourhood_top'] = df['neighbourhood_cleansed'].where(
    df['neighbourhood_cleansed'].isin(top_neigh), 'Other'
)

# Convert to dummy variables, drop first to avoid multicollinearity
df = pd.get_dummies(df, columns=['neighbourhood_top'], drop_first=True)

# Check the first few rows to confirm
print(df[['log_revenue', 'log_price', 'log_reviews']])
```

```python
# List of neighborhood dummy columns
neigh_cols = [
    'neighbourhood_top_Brighton',
    'neighbourhood_top_Dorchester',
    'neighbourhood_top_Downtown',
    'neighbourhood_top_Other',
    'neighbourhood_top_Roxbury'
]

# Convert boolean to 0/1
for col in neigh_cols:
    df[col] = df[col].astype(int)

# Verify
print(df[neigh_cols].dtypes)
print(df[neigh_cols].head())
```

```python
#recheck remaining columns
df.info()
```

```python
#Drop remaining, unnecessary object columns
cols_to_drop = ['neighbourhood_cleansed', 'property_type', 'room_type']

df = df.drop(columns=cols_to_drop)

# Verify they were removed
df.info()
```

```python
import statsmodels.api as sm

# --- DEFINE CONTROL VARIABLES ---
controls = [
    'host_is_superhost',
    'host_response_rate',
    'instant_bookable',
    'log_reviews',
    'review_scores_rating'
]

# --- Fix neighborhood dummies ---
neigh_cols = [c for c in df.columns if c.startswith('neighbourhood_top_')]
for col in neigh_cols:
    df[col] = df[col].astype(int)

# Drop the first dummy to serve as baseline
controls += neigh_cols[1:]

# Ensure all controls are numeric
for col in controls:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# --- MODEL 1: Controls only ---
X1 = df[controls].dropna()
y1 = df['log_revenue'].loc[X1.index]
X1 = sm.add_constant(X1)
model1 = sm.OLS(y1, X1).fit(cov_type='HC3')
print("=== Model 1 (Controls Only) ===")
print(model1.summary())

# --- MODEL 2: Controls + main IVs ---
main_IVs = ['log_price', 'minimum_nights', 'accommodates']
X2 = df[controls + main_IVs].dropna()
y2 = df['log_revenue'].loc[X2.index]
X2 = sm.add_constant(X2)
model2 = sm.OLS(y2, X2).fit(cov_type='HC3')
print("\n=== Model 2 (Controls + Main IVs) ===")
print(model2.summary())

# --- MODEL 3: Add interaction terms ---
# Original interactions
df['price_x_accom'] = df['log_price'] * df['accommodates']
df['min_nights_x_accom'] = df['minimum_nights'] * df['accommodates']

# Additional neighborhood interactions
df['price_x_downtown'] = df['log_price'] * df['neighbourhood_top_Downtown']
df['price_x_roxbury'] = df['log_price'] * df['neighbourhood_top_Roxbury']
df['min_nights_x_downtown'] = df['minimum_nights'] * df['neighbourhood_top_Downtown']
df['min_nights_x_roxbury'] = df['minimum_nights'] * df['neighbourhood_top_Roxbury']
df['accom_x_downtown'] = df['accommodates'] * df['neighbourhood_top_Downtown']
df['accom_x_roxbury'] = df['accommodates'] * df['neighbourhood_top_Roxbury']

interactions = [
    'price_x_accom', 'min_nights_x_accom',
    'price_x_downtown', 'price_x_roxbury',
    'min_nights_x_downtown', 'min_nights_x_roxbury',
    'accom_x_downtown', 'accom_x_roxbury'
]

X3 = df[controls + main_IVs + interactions].dropna()
y3 = df['log_revenue'].loc[X3.index]
X3 = sm.add_constant(X3)
model3 = sm.OLS(y3, X3).fit(cov_type='HC3')
print("\n=== Model 3 (Main IVs + Interactions) ===")
print(model3.summary())
```
